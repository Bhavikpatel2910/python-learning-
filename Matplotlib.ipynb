```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
x
y=x**2
y
```

```python
plt.plot(x,y)
plt.title('Bhavik')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x*4,y)
plt.subplot(2,2,2)
plt.subplot(2,2,1)
```
